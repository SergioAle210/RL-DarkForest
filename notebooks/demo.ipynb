# Reproducimos nuestros episodios

Leemos episodios guardados. Esta vista no ejecuta el entorno ni entrena el modelo.
Primero generamos una corrida según [la guía](../docs/EPISODIOS.md), o recibimos
la carpeta completa de un compañero dentro de runs/config/seed.

```python
from pathlib import Path

import matplotlib.pyplot as plt
from IPython.display import HTML, display

from analysis.metrics import summarize
from darkforest.trajectories import load_episode
from viz.galaxy import episode_animation

ROOT = next(
    path for path in (Path.cwd(), *Path.cwd().parents) if (path / "pyproject.toml").is_file()
)


episodes = sorted((ROOT / "runs").glob("*/*/trajectories/episode_*.npz"))
if not episodes:
    raise FileNotFoundError("Primero generamos o copiamos una corrida con trayectorias")
for index, path in enumerate(episodes):
    print(index, path.relative_to(ROOT))
```

## Seleccionamos un episodio

Elegimos el índice de la lista anterior. El JSON asociado indica configuración, seed
y política; el NPZ conserva T acciones y T+1 estados para reproducir la secuencia.

```python
episode_index = 0
trajectory, metadata = load_episode(episodes[episode_index])
print("Política:", metadata["policy"], "Seed:", metadata["seed"])
print(summarize(trajectory))
```

## Observamos la galaxia y sus recursos

Los puntos representan civilizaciones vivas; las cruces representan eliminadas. Los anillos
muestran señales, los triángulos proyectiles y el área roja un estallido. La posición de
cada civilización permanece fija, como plantea la propuesta. La vista global usa datos
de diagnóstico guardados; la política solo recibe observaciones parciales.

```python
animation = episode_animation(trajectory, metadata, fps=8)
display(HTML(animation.to_jshtml()))
plt.close(animation._fig)
```

Podemos cambiar episode_index para comparar episodios de dos y cuatro civilizaciones
o políticas iniciales y entrenadas. Conservamos las corridas fuera de Git y limpiamos
las salidas del notebook antes de hacer commit.
